```python
from rouge import Rouge
from nltk.translate.bleu_score import sentence_bleu
from sentence_transformers import SentenceTransformer, util
import numpy as np

rouge = Rouge()
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
def rouge_l_score(pred, ref):
    try:
        return rouge.get_scores(pred, ref)[0]["rouge-l"]["f"]
    except:
        return 0.0
```

```python
def bleu_score(pred, ref):
    try:
        return sentence_bleu([ref.split()], pred.split())
    except:
        return 0.0
```

```python
def embedding_similarity(pred, ref):
    try:
        e1 = embedding_model.encode(pred)
        e2 = embedding_model.encode(ref)
        return float(util.cos_sim(e1, e2))
    except:
        return 0.0
```

```python
def factual_accuracy(pred, ref):
    pred_set = set(pred.lower().split())
    ref_set = set(ref.lower().split())
    return len(pred_set & ref_set) / max(1, len(ref_set))
```

```python
def hallucination_penalty(pred, ref):
    pred_set = set(pred.lower().split())
    ref_set = set(ref.lower().split())
    hallucinated = pred_set - ref_set
    return -len(hallucinated) / max(1, len(pred_set))
```

```python
def rag_retrieval_precision(docs, expected):
    if not docs:
        return 0.0
  
    score = 0
    for d in docs:
        if expected.lower() in d.lower():
            score += 1

    return score / len(docs)
```
