```python
import json

with open("dataset_stage2_wikipedia_augmented.json") as f:
    wikipedia_data = json.load(f)

with open("dataset_stage1_base_symptoms.json") as f:
    symptoms_data = json.load(f)

with open("dataset_stage3_qa_filtered.json") as f:
    qa_data = json.load(f)
```

```python
disease_map = {entry["disease"].lower(): entry for entry in wikipedia_data}
```

```python
for qa in qa_data:
    question = qa["question"]
    answer = qa["answer"]

    for disease_name, disease_entry in disease_map.items():
        if disease_name in question.lower():
            if "qa_pairs" not in disease_entry:
                disease_entry["qa_pairs"] = []
            disease_entry["qa_pairs"].append({
                "question": question,
                "answer": answer
            })
```

```python
for disease_name, entry in disease_map.items():
    entry["emotional_support"] = (
        f"I know dealing with {entry['disease']} can feel stressful, but you are not alone. "
        "With proper care and guidance, most people manage it successfully. You're doing the right thing by learning more."
    )
```

```python
final_dataset = list(disease_map.values())
final_dataset = [d for d in final_dataset if d["summary"]]
```

```python
with open("MedicalLLM_final_dataset.json", "w") as f:
    json.dump(final_dataset, f, indent=4)
```
