```python
import pandas as pd

df = pd.read_csv("disease-symptom-description-dataset/dataset.csv")

print(df.head())
print(df.info())
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.head()
```

```python
processed = []

symptom_cols = [col for col in df.columns if col.startswith("symptom_")]

for _, row in df.iterrows():
    disease = row["disease"].strip()

    symptoms = []
    for col in symptom_cols:
        val = row[col]
        if isinstance(val, str) and val.strip():
            symptoms.append(val.strip())

    entry = {
        "disease": disease,
        "symptoms": symptoms,
        "summary": "",
        "cause": "",
        "treatments": [],
        "emotional_support": "",
        "qa_pairs": [],

        "sources": ["kaggle_disease_symptoms_patient_profile"]
    }

    processed.append(entry)

len(processed)
```

```python
unique_data = {}
for entry in processed:
    d = entry["disease"].lower()
    if d not in unique_data:
        unique_data[d] = entry

dataset = list(unique_data.values())
len(dataset)
```

```python
import json

with open("dataset_stage1_base_symptoms.json", "w") as f:
    json.dump(dataset, f, indent=4)
```
