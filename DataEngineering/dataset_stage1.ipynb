```python
import pandas as pd

# Change the path to where Kaggle downloaded the file
df = pd.read_csv("chicago-disease-symptoms-dataset/disease_symptom.csv")

df.head()
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.head()
```

```python
processed = []

for _, row in df.iterrows():
    disease = row["disease"]
    
    # symptom split (depends on dataset)
    symptoms = row["symptoms"].split(",") if isinstance(row["symptoms"], str) else []

    entry = {
        "disease": disease.strip(),
        "symptoms": [s.strip() for s in symptoms if s.strip()],
        "summary": "",
        "cause": "",
        "treatments": [],
        "emotional_support": "",
        "qa_pairs": [],
        "sources": ["kaggle_symptom_dataset"]
    }

    processed.append(entry)

len(processed)
```

```python
unique_data = {}
for entry in processed:
    d = entry["disease"].lower()
    if d not in unique_data:
        unique_data[d] = entry

dataset = list(unique_data.values())
len(dataset)
```

```python
import json

with open("dataset_stage1_base_symptoms.json", "w") as f:
    json.dump(dataset, f, indent=4)
```
