```python
import json

with open("MedicalLLM_final_dataset.json") as f:
    data = json.load(f)
```

```python
sft_rows = []

for d in data:
    instruction = f"Provide medical information about {d['disease']}."
    output = (
        f"Summary: {d['summary']}\n"
        f"Cause: {d['cause']}\n"
        f"Symptoms: {', '.join(d['symptoms'])}\n"
        f"Treatments: {', '.join(d['treatments'])}\n"
        f"Emotional Support: {d['emotional_support']}"
    )

    sft_rows.append({
        "instruction": instruction,
        "input": "",
        "output": output
    })
```

```python
for d in data:
    for symptom in d["symptoms"]:
        instruction = f"Identify possible diseases for the symptom: {symptom}"
        output = f"{d['disease']}: Possible cause due to presence of {symptom}."

        sft_rows.append({
            "instruction": instruction,
            "input": "",
            "output": output
        })
```

```python
for d in data:
    for qa in d["qa_pairs"]:
        sft_rows.append({
            "instruction": "Answer the medical question.",
            "input": qa["question"],
            "output": qa["answer"]
        })
```

```python
import random
random.shuffle(sft_rows)

train_size = int(0.9 * len(sft_rows))
train_data = sft_rows[:train_size]
val_data = sft_rows[train_size:]

with open("sft_train.json", "w") as f:
    json.dump(train_data, f, indent=4)
with open("sft_val.json", "w") as f:
    json.dump(val_data, f, indent=4)
```
